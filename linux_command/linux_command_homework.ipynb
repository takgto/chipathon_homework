# 第2章　ディレクトリを作って移動し、ファイルを整理する

## 解説

演習用に、下のようなフォルダ構成を自分のPCに作ります。

```
practice/
├── data/
└── src/
```

そのうえで、ファイルを作ったりコピー・移動したりして整理します。

- フォルダ作成は `mkdir`（make directory）、空ファイル作成は `touch`。
- コピーは `cp`（元が残る）、移動・改名は `mv`（元が無くなる）、削除は `rm`。
- `..` は「1つ上の階層」、`~` はホームディレクトリを指す。

---

## 演習

### 問4

ホームディレクトリに `practice` フォルダを作り、その中に `data` と `src` の2つのフォルダを作るには？（コマンドは何回かに分けてもよい）

### 問5

`practice/src` に移動して、`hello.txt` という空のファイルを新規作成するには？

### 問6

その `hello.txt` を、1つ上の階層の `data` フォルダにコピーするには？

### 問7

`src` 側の `hello.txt` の名前を `memo.txt` に変更するには？

### 問8

最後に `memo.txt` を削除するには？

**ヒント**：フォルダ作成は `make directory` の略。空ファイル作成は `touch`。コピーは `cp`、移動・改名は `mv`、削除は `rm`。`..` は「1つ上の階層」を指す。

## 演習の解答

### 問4

```bash
cd ~                     # ホームに移動
mkdir practice           # practice フォルダを作成
cd practice
mkdir data src           # data と src をまとめて作成（空白区切りで複数指定）
```

- `mkdir -p practice/data` のように `-p` を付けると、`practice` が無くても親フォルダごと一気に作れます。

### 問5

```bash
cd src                   # src に移動
touch hello.txt          # 空ファイルを作成
```

### 問6

```bash
cp hello.txt ../data/    # 1つ上(..)の data/ へコピー
```

### 問7

```bash
mv hello.txt memo.txt    # mv は「移動」だが、同じ場所なら「名前変更」になる
```

- `cp`（コピー）は元が残る、`mv`（移動/改名）は元が無くなる、と覚えると区別しやすいです。

### 問8

```bash
rm memo.txt              # 削除
```

- `rm` は **ごみ箱を経由せず即削除** され、元に戻せません。消す前に必ず `ls` で対象を確認してください。フォルダごと消すときは `rm -r フォルダ名`。

````python
%%bash
# 第2章の確認（各手順のあとに ls で中身を表示して変化を追う）
cd ~
mkdir -p practice/data practice/src
cd practice/src
touch hello.txt
echo "--- src (touch 後) ---";      ls
cp hello.txt ../data/
echo "--- data (cp 後) ---";        ls ../data
mv hello.txt memo.txt
echo "--- src (mv 後) ---";         ls
rm memo.txt
echo "--- src (rm 後・空) ---";     ls
````

# 第3章　シェルスクリプトを作って実行する（権限の演習）

## 解説

本番では転送した実行ファイルを `./プログラム名` で動かします。その演習として、自分で簡単なスクリプトを作り、実行できる状態にして動かしてみます。

作りたい `hello.sh` の中身は次の **2行** です。

```bash
#!/bin/bash
echo "Hello, KV260!"
```

- `>` は「上書き保存」、`>>` は「追記」（リダイレクト）。
- 実行権限の付与は `chmod`。`+x` で実行権限（execute）を追加。
- 実行するときは先頭に `./` が必要。

---

## 演習

### 問9

この2行が書かれた `hello.sh` を、エディタを使わず、`echo` とリダイレクト（`>` と `>>`）だけで作るには？

### 問10

作った `hello.sh` を `./hello.sh` で実行しようとすると `Permission denied`（権限がありません）と出た。実行権限を付けるコマンドは？

### 問11

権限を付けたあと、改めて実行するには？

**ヒント**：`>` は「上書き保存」、`>>` は「追記」。実行権限の付与は `chmod`、`+x` で実行権限(execute)を追加。実行するときは先頭に `./` が必要。

## 演習の解答

### 問9

```bash
# echo の出力をファイルに書き込む（> で新規、>> で追記）
echo '#!/bin/bash' > hello.sh
echo 'echo "Hello, KV260!"' >> hello.sh
cat hello.sh                 # 中身を確認（cat はファイル内容を表示）
```

### 問10

```bash
chmod +x hello.sh            # 実行権限(x)を付与
```

- `ls -l hello.sh` で権限を確認できます。`-rw-r--r--`（xなし＝実行不可）が、`chmod +x` 後に `-rwxr-xr-x`（xあり＝実行可）に変わるのを見てください。`r`=読み取り、`w`=書き込み、`x`=実行。

### 問11

```bash
./hello.sh                   # → Hello, KV260! と表示される
```

- なぜ `./` が必要か：Linuxは「今いるフォルダ」を自動では探しません。`hello.sh` だけだと「そんなコマンドは無い」と言われます。`./` で「このフォルダの中の」と明示します。
- 本番のボードでは、デバイス操作などで管理者権限が必要な場面があり、その際はコマンドの先頭に `sudo` を付けます（例：`sudo ./run_yolo`）。`sudo` は強力なので、何をするコマンドか分かったうえで使ってください。

````python
%%bash
# 第3章の確認：権限が無いと Permission denied、chmod +x 後は実行できる
cd ~/practice/src
echo '#!/bin/bash' > hello.sh
echo 'echo "Hello, KV260!"' >> hello.sh
echo "--- cat hello.sh ---";  cat hello.sh
echo "--- chmod 前 ---";      ls -l hello.sh
./hello.sh || echo "(実行できない: Permission denied)"
chmod +x hello.sh
echo "--- chmod 後 ---";      ls -l hello.sh
./hello.sh
````

# 第4章　ファイルの圧縮・展開とディスク容量の確認

## 解説

本番のKV260はSDカードの容量に限りがあります。大きな学習データを圧縮ファイル（`.tar.gz`）でやり取りし、空き容量を気にしながら作業します。その演習を手元で行います。

- `tar` コマンドを使う。固める＝`-c`(create)、展開＝`-x`(extract)、gzip圧縮＝`-z`、ファイル指定＝`-f`、経過表示＝`-v`。
- 空き容量は `df`、フォルダ使用量は `du`。`-h`(human-readable) で単位が読みやすくなる。

---

## 演習

### 問12

第2章で作った `practice` フォルダを丸ごと `practice.tar.gz` という圧縮ファイルに固めるには？

### 問13

その `practice.tar.gz` を展開（解凍）するには？

### 問14

ディスク全体の **空き容量** を、人間が読みやすい単位（GB/MB）で表示するには？

### 問15

`practice` フォルダが **どれくらいの容量を使っているか** を調べるには？

**ヒント**：`tar` コマンドを使う。固める＝`-c`(create)、展開＝`-x`(extract)、gzip圧縮＝`-z`、ファイル指定＝`-f`、経過表示＝`-v`。空き容量は `df`、フォルダ使用量は `du`。`-h`(human-readable)で単位が読みやすくなる。

## 演習の解答

### 問12

```bash
tar -czvf practice.tar.gz practice     # c=作成 z=gzip v=経過 f=ファイル名
```

- `tar` のオプションは `-czvf` のように **まとめて書ける** のが定番です。`-f` の直後にファイル名を書く点に注意。

### 問13

```bash
tar -xzvf practice.tar.gz              # x=展開 z=gzip v=経過 f=ファイル名
```

- `c`(固める)と `x`(展開)を入れ替えるだけで圧縮・解凍を切り替えられます。

### 問14

```bash
df -h                                  # ディスクの空き容量（-h で GB/MB 表記）
```

- `df -h` の `Avail`（利用可能）列が空き容量です。本番ではここがSDカードの残量で、`100%` に近いと作業中にエラーになります。

### 問15

```bash
du -sh practice                        # -s=合計のみ -h=読みやすい単位
```

- `du -sh *` とすると、フォルダ内の各項目の容量を一覧でき、「何が容量を食っているか」を特定できます。

````python
%%bash
# 第4章の確認
cd ~
echo "--- 問12. 圧縮 ---";        tar -czvf practice.tar.gz practice
echo "--- 問13. 展開 ---";        tar -xzvf practice.tar.gz
echo "--- 問14. 空き容量 ---";    df -h /
echo "--- 問15. 使用容量 ---";    du -sh practice
````

# 第5章　SSHとSCPを実際に試す（自分のPCを"サーバー役"にする実験）

## 解説

本番では、手元のPCから **ネットワーク越しにKV260を操作** します。そのための2大コマンドが `ssh`（ボードにログインする）と `scp`（ファイルを転送する）です。実機はまだ無いので、**自分のPC自身を相手サーバーに見立てて** 演習します。

> **考え方**：`localhost`（＝自分自身、IPアドレスでいうと `127.0.0.1`）を、本番のKV260のIPアドレスの「代役」にします。コマンドの形は本番とまったく同じで、接続先が「自分自身」になるだけです。

> **注意**：この章はColabでは試せません。自分のPCのターミナルで行ってください。

### 準備：自分のPCにSSHサーバーを立てる

`ssh` で「ログインされる側（サーバー）」になるには、SSHサーバー（sshd）を動かす必要があります。下の **自分のPCの種類に合った手順** を行ってください。（この準備手順は答えではないので、そのまま実行してください）

**Windowsの人（WSL2 / Ubuntu のターミナルで実行）**

```bash

sudo apt update
sudo apt install -y openssh-server

# 2. SSHサーバーを起動する（WSL2では起動のたびに必要なことが多い）

# 3. 起動できたか確認（Active: active (running) と出ればOK）
```

- 途中でパスワードを聞かれたら、WSL2のUbuntu初回設定時に決めた **自分のパスワード** を入力します（画面には何も表示されませんが入力できています）。
- `sudo service ssh start` を実行しても **特にメッセージは表示されません**（無言で起動します）。`sudo service ssh status` を実行して `Active: active (running)` と表示されれば起動成功です。

**Macの人（GUIの設定画面で操作）**

1. 画面左上のアップルメニュー →「システム設定」を開く。
2. 「一般」→「共有」を選ぶ。

## 演習の解答

### 問16

```bash
ssh あなたのユーザー名@localhost
#    初回は "Are you sure...?" と聞かれるので yes と入力
#    その後パスワードを入力するとログイン完了
```

- 初回確認には `yes` と入力します（接続先を信頼する、という確認です）。
- **「ログインしても何も変わらないように見える」のは正常です。** `localhost` は **自分自身のMac/PC** なので、接続先が「同じマシン」です。ホスト名もホームディレクトリもファイルもすべて同じなので、見た目に変わるところがありません。本番のKV260では `you@PC` → `root@kria` のように **別マシンに入るため行頭の表示（プロンプト）がはっきり変わります**。今どちらのマシンを操作しているか常に確認する癖をつけてください。
- **本当にSSH接続できているかの確認方法**：ログイン後に `echo $SSH_CONNECTION` を実行します。
  - 何か表示されれば（例：`::1 49492 ::1 22`）→ **SSH接続中**（成功）。
  - 何も表示されず空っぽなら → まだSSHに入れていません（ローカルのまま）。
  - `exit` で抜けたあと同じコマンドを打つと、今度は何も表示されません。これも「SSHの層を抜けた」証拠です。
- **`echo $SSH_CONNECTION` の表示（例 `::1 49492 ::1 22`）の意味**：左から「接続元IP / 接続元ポート / 接続先IP / 接続先ポート」です。`::1` はIPv6版のループバックアドレス（IPv4の `127.0.0.1` に相当）で「自分自身」を指します。`49492` はOSが自動で割り当てる一時的なポート（毎回変わります）、`22` はSSHの標準ポートです。つまり「自分自身から自分自身のSSHポートへ繋いでいる」状態を表します。本番ではこの接続先IPが `192.168.1.100`（KV260）のような実際のアドレスになります。

- `scp` は必ず **「コピー元」「コピー先」の順**。`:` の右側が相手側の場所です。本番では `localhost` の部分が `root@192.168.1.100`（KV260）に変わるだけで、コマンドの形は同じです。
- **本番との対応**：
  - ログイン　`ssh root@192.168.1.100`
  - 送る　　　`scp model.xmodel root@192.168.1.100:~/work`
  - 受け取る　`scp root@192.168.1.100:~/work/result.jpg .`

  今回の実験で `localhost` を相手にした形が、そのまま当日のボード操作になります。

# 第6章　VS Codeでリモートのファイルを編集する（VS Code Remote-SSH）

## 解説

## 演習の解答

### 問20

VS Code画面の **左下にある緑色の `><` マーク** をクリック →「**Connect to Host...**」を選ぶ。一覧に `localhost` が無ければ「**+ Add New SSH Host...**」を選び、`ssh あなたのユーザー名@localhost` と入力して登録してから接続します。（`F1` または `Ctrl/Cmd + Shift + P` でコマンドパレットを開き、`Remote-SSH: Connect to Host...` と打ってもOK）

- 初回は接続先の確認（fingerprint）を聞かれるので「**Continue**」、続いてパスワードを入力します。
- 接続が成功すると、新しいVS Codeウィンドウが開き、左下の表示が `SSH: localhost` に変わります。

### 問22

画面 **左下の緑色のインジケータ** に `SSH: localhost` と表示されていれば、リモート（接続先）に繋がっている状態です。ローカルで作業中はこの表示が出ません。内蔵ターミナルのプロンプトや、エクスプローラーのパス（`/home/...`）でも接続先を確認できます。

- **本番との対応**：左下のインジケータから接続するホスト名を `localhost` → `root@192.168.1.100`（KV260）に変えるだけで、まったく同じ手順でボード上の `*.cpp` を編集できます。編集→内蔵ターミナルでビルド→実行、という流れも本番でそのまま使えます。
- **うまく接続できないとき**：第5章のSSHサーバーが起動しているか確認してください（WSL2: `sudo service ssh start`）。それでも繋がらない場合はポートが競合している可能性があるので、担当者に相談してください。

**補足：接続を簡単にする（`~/.ssh/config` にエイリアスを登録）**

毎回 `root@192.168.1.100` と入力する代わりに、SSH設定ファイルにホストの別名（エイリアス）を書いておくと便利です。

- Windows：`C:\Users\<ユーザー名>\.ssh\config`
- Mac：`~/.ssh/config`（＝ `/Users/<ユーザー名>/.ssh/config`）

に次のように記述します（`Host` の行が別名）。

```
Host kv260
    HostName 192.168.1.100
    User root
```

こう書いておくと、VS CodeのRemote-SSHで「Connect to Host...」を選んだとき、ホストの一覧に `kv260` が現れます。それを選ぶだけで `kv260`（＝ `root@192.168.1.100`）へRemote-SSH接続できます。ターミナルからも `ssh kv260` だけで接続できるようになり、毎回 `root@192.168.1.100` と入力する必要がなくなります。
